If you have a list l, these are the options:

- l[start:stop]  # items start through stop-1
- l[start:]      # items start through the rest of the array
- l[:stop]       # items from the beginning through stop-1
- l[:]           # a copy of the whole array

## Event-driven and Sequential

In this week's workshop, we will go again over all the concepts and practices we have just discussed.  But before that, we need to introduce a few more aspects of Python, the programming language, which you might need in data analysis tasks, but they are much more common if you want to program say an app. The first topic is control structures.

Control structures execute a piece of code based on a condition. You can recognise them in most programming languages by the keyword `if`. In Python, they look like the cell below. Run the next cell, please.

````python
#Keep cell
medium = 'Social Media'

if (medium == 'Social Media'):
    print('Showing Social Media information')
````

The first statement creates a variable `medium` with 'Social Media' as a value. The control structure starts with `if`, then uses a condition to evaluate to a Boolean (`medium == 'Social Media'`), followed by a colon. The statement to execute follows in the next line.  `print('Showing Social Media information')` will print out if the condition is met.

Important: Observe the indentation, which is a way of telling Python that a group of statements belongs together as a block. Blocks can be regarded as the grouping of statements for a specific purpose. Python really does not like if you mess with its indentation. Check https://www.dummies.com/article/technology/programming-web-design/python/how-to-indent-and-dedent-your-python-code-139545/.

Now, also assign ```num_views = 1001```.

Let's try to confirm whether we are popular with ```if num_views > 1000``` then print 'You are very popular!'

Now, using the `num_views` variable, write an if statement that prints 'You are very popular!' if `num_views` is greater than 1000. You can do it ...

The keyword 'and' stands for a logical and, while 'or' stands for the logical or in Python.

Explain the results!

`num_views > 1000` and `medium == 'Social Media'` means that the if condition is met and 'You are popular on Social Media!' is printed.

Finally, you can also tell the computer to do other work if the condition is not fulfilled. You need to use the keyword `else`. Run the cell below and observe the indentation ...

````python
#Keep cell

if (num_views > 1000) and (medium == 'Social Media'):
    print('Condition met.')
    print('You are popular on Social Media!')
else:
    print('Condition not met.')
    print('Try to be more visible!')
````

Let's bring loops and conditions together and compare `friends_list` and `friends_array`.

We want to print out 'The lists have it!' each time the `friends_list` element at position `i` is larger than or equal to the `friends_array` element at position `i`. Otherwise, we print 'The arrays have it!'.

Take a moment to think how to do it. You need both for-loops and if-statements.

