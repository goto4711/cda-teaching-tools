**Before you run anything:** Runtime → Change runtime type → **T4 GPU** → Save.
Then Runtime → **Run all**, and leave it. Loading the model takes one to two minutes; after that each picture takes about a second.

The model is [SD-Turbo](https://huggingface.co/stabilityai/sd-turbo) from Hugging Face: a small, fast version of Stable Diffusion that makes a picture in a single step.

```python
# 1. Is there a GPU? (T4 is the free one)
import torch
if not torch.cuda.is_available():
    raise SystemExit("No GPU: choose Runtime → Change runtime type → T4 GPU, then Run all again.")
print("GPU:", torch.cuda.get_device_name(0))
```

```python
# 2. The library that runs diffusion models (Colab may already have it)
!pip install -q diffusers
```

```python
# 3. Download the model from Hugging Face: about 2.5 GB of numbers
import time
from diffusers import AutoPipelineForText2Image

t = time.time()
pipe = AutoPipelineForText2Image.from_pretrained(
    "stabilityai/sd-turbo", torch_dtype=torch.float16, variant="fp16"
).to("cuda")
print(f"Model ready after {time.time() - t:.0f} seconds")
```

```python
# 4. Four sentences, four pictures
from diffusers.utils import make_image_grid

prompts = [
    "an astronaut riding a green horse",
    "a museum gallery full of teenagers looking at their phones, oil painting",
    "the canals of Amsterdam in the year 2200, watercolour",
    "a library where the books are made of data, photograph",
]
t = time.time()
images = [pipe(prompt=p, num_inference_steps=1, guidance_scale=0.0).images[0] for p in prompts]
print(f"{len(images)} pictures in {time.time() - t:.1f} seconds")
make_image_grid(images, rows=1, cols=4)
```
